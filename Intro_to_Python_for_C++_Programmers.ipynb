# Python vs. C++: Key Differences

## Interpreted vs. Compiled Languages

*   **Compiled Languages (like C++)**: Code written in a compiled language is first translated into machine-readable instructions (machine code) by a program called a compiler. This compilation step happens once, before the program is run. The resulting executable file can then be run directly by the computer's processor. This usually leads to faster execution speeds.

*   **Interpreted Languages (like Python)**: Code written in an interpreted language is executed line by line at runtime by an interpreter. The interpreter reads the source code, translates it into an intermediate form (like bytecode), and then executes it. There isn't a separate, explicit compilation step that produces a standalone executable. This often allows for quicker development cycles and greater flexibility.

| **Execution**        | Interpreted                              | Compiled                                        |
| **Typing**           | Dynamically typed, strong                | Statically typed, strong                        |
| **Memory Management**| Automatic (garbage collection)           | Manual (pointers, `new`/`delete`)               |

## Python Objects: Messages and Types

In Python, *everything is an object*. This is a fundamental concept. An object can be thought of as a self-contained entity that has both data (attributes) and behavior (methods). You interact with objects by "sending them messages," which in Python means calling their methods or accessing their attributes using dot notation (`object.method()` or `object.attribute`).

### Message-Sending Syntax

The syntax for "sending a message" to an object (calling a method) is straightforward: `object_name.method_name(arguments)`. When you call a method on an object, you are asking that specific object to perform an action using its internal data.

```python
my_string = "hello python"
my_list = [1, 2, 3]
my_number = 10

# Sending a 'message' (calling a method) to a string object
print(f"Uppercase string: {my_string.upper()}")

# Sending a 'message' (calling a method) to a list object
my_list.append(4)
print(f"List after appending: {my_list}")

# Integers are also objects with methods (though less frequently used directly)
print(f"10 to binary: {my_number.bit_length()} bits")
```

### Everything is an Object (and has a Type)

In Python, integers, floats, strings, lists, dictionaries, functions, and even classes themselves are objects. Each object is an instance of a particular type (or class). You can inspect an object's type using the `type()` function.

```python
a_string = "Python"
a_integer = 42
a_list = [10, 20]
a_function = len

print(f"Type of '{a_string}': {type(a_string)}")
print(f"Type of {a_integer}: {type(a_integer)}")
print(f"Type of {a_list}: {type(a_list)}")
print(f"Type of {a_function}: {type(a_function)}")

# Even a class itself is an object of type 'type'
print(f"Type of the list class: {type(list)}")
```

### Common Built-in Types

Python provides several fundamental built-in types that are commonly used:

*   **Numeric Types**: `int` (integers), `float` (floating-point numbers), `complex` (complex numbers).
*   **Sequence Types**: `str` (strings - immutable), `list` (lists - mutable), `tuple` (tuples - immutable).
*   **Mapping Type**: `dict` (dictionaries - mutable, key-value pairs).
*   **Set Types**: `set` (mutable, unordered collections of unique elements), `frozenset` (immutable version of set).
*   **Boolean Type**: `bool` (True/False).
*   **None Type**: `NoneType` (represents the absence of a value).

All of these are objects, instances of their respective classes, demonstrating Python's consistent object model.

## Loose (Dynamic) Typing vs. Strict (Static) Typing

One significant difference between Python and C++ lies in their typing systems:

*   **Python (Dynamically Typed)**: In Python, you do not need to explicitly declare the type of a variable when you create it. The type is determined at *runtime* based on the value assigned to it. A single variable can even hold values of different types throughout its lifetime. This offers flexibility and often leads to more concise code.

*   **C++ (Statically Typed)**: In C++, every variable must have its type explicitly declared *before* it can be used. This type declaration is checked at *compile-time*. Once a variable is declared with a certain type, it can only hold values of that type (or types convertible to it). This strictness helps catch type-related errors early in the development process and can lead to more optimized code, but it requires more verbose syntax.

```python
# Python's dynamic typing in action

x = 10          # x is an integer
print(f"Value of x: {x}, Type of x: {type(x)}")

x = "hello"     # Now x is a string
print(f"Value of x: {x}, Type of x: {type(x)}")

x = [1, 2, 3]   # Now x is a list
print(f"Value of x: {x}, Type of x: {type(x)}")

# In C++, this would result in a compile-time error:
# int x = 10;
# x = "hello"; # Error: cannot convert 'const char*' to 'int'
```

## Python Strings: Immutable Sequences

Strings in Python are sequences of characters, but a crucial characteristic is their *immutability*. This means that once a string is created, its contents cannot be changed. Any operation that appears to modify a string, such as concatenation or slicing, actually creates a *new* string object in memory, leaving the original string untouched. Understanding immutability is an important concept in Python programming.

```python
s = "hello"
print(f"Original string: {s}")

# Attempting to modify a character will result in an error
try:
  s[0] = 'H'
except TypeError as e:
  print(f"Error trying to modify string character: {e}")

# String operations like concatenation create new strings
s_new = s + " world"
print(f"New string after concatenation: {s_new}")
print(f"Original string remains: {s}")

# Even slicing creates a new string
sub_s = s[1:4]
print(f"Substring from slicing: {sub_s}")
```

### String Slicing

String slicing allows you to extract portions (substrings) of a string. This operation also produces a *new* string. Here are various examples demonstrating string slicing:

```python
s = "banana"

# Get the first character
print(f"s[0]: {s[0]}")

# Get characters from index 1 to 4 (exclusive)
print(f"s[1:5]: {s[1:5]}")

# Get characters from the beginning to index 3 (exclusive)
print(f"s[:4]: {s[:4]}")

# Get characters from index 3 to the end
print(f"s[3:]: {s[3:]}")

# Get characters from the second to last to the end
print(f"s[-2:]: {s[-2:]}")

# Get characters in reverse order
print(f"s[::-1]: {s[::-1]}")

# Get every second character
print(f"s[::2]: {s[::2]}")
```

## Python Lists: Mutable Sequences

In stark contrast to strings, Python lists are *mutable*. This means their contents can be changed after they are created. You can add, remove, or modify elements directly within the same list object. This mutability is a key difference from strings and offers flexibility in data manipulation. Lists are ordered collections of items, and they can contain items of different data types.

```python
# Define a list of names
names = ["Ellie", "Elijah", "Matt"]

# Print the type and length of the list
print(f"Type of 'names': {type(names)}")
print(f"Length of 'names': {len(names)}")
```

### Iterating Through a List

We can loop through each item in a list and perform an action, such as converting each name to uppercase.

```python
print("Names in uppercase:")
for name in names:
  print(name.upper())
```

### Modifying a List During Iteration

It's generally not recommended to modify a list (add or remove items) while iterating over it directly, as it can lead to unexpected behavior. A common pattern is to iterate over a *copy* of the list if you need to remove elements from the original. This ensures the iteration process isn't disrupted by changes to the list's size or indexing. This example demonstrates removing names that contain the letter 'e' (case-insensitive).

```python
print("\nOriginal names list:", names)

# Iterate over a copy of the list to avoid issues when removing elements
# If we iterate over `names` directly and remove an element, the indices shift,
# and we might skip elements or encounter an IndexError.
initial_names = list(names) # Create a copy for safe iteration

for name in initial_names:
  if "e" in name.lower():
    names.remove(name)

print("Names after removing those with 'e':", names)
```

## Python Dictionaries: Key-Value Mappings

Dictionaries are another fundamental built-in data type in Python. They are unordered collections of key-value pairs. Dictionaries are *mutable*, meaning their contents can be changed after creation. Keys must be unique and immutable types (like strings, numbers, or tuples), while values can be any data type, making them very flexible for storing structured data.

```python
# Create a dictionary
person = {
  "name": "Alice",
  "age": 30,
  "city": "New York"
}
print(f"Original dictionary: {person}")

# Accessing values by key
print(f"Name: {person['name']}")
print(f"Age: {person.get('age')}") # Using .get() for safer access

# Modifying a value
person['age'] = 31
print(f"Modified age: {person['age']}")

# Adding a new key-value pair
person['occupation'] = "Engineer"
print(f"Dictionary after adding occupation: {person}")

# Removing a key-value pair
del person['city']
print(f"Dictionary after removing city: {person}")
```

## Python Classes: Object-Oriented Programming

Classes provide a blueprint for creating objects (instances) that encapsulate their own data (attributes) and functionality (methods). This allows for organizing code into reusable and modular components, which is a core concept of Object-Oriented Programming (OOP).

### The `Student` Class

This `Student` class demonstrates how to define a custom object. It includes:

*   **`__init__(self, name, age)`**: This is the *constructor* or initializer method. It's automatically called when a new `Student` object is created. The `self` parameter is a convention (but not a keyword) that refers to the instance of the class itself, and it must be the first parameter of any instance method. Python implicitly passes the instance when a method is called.
*   **Attributes**: `name`, `age`, and `purchases` store data specific to each student object.
*   **Methods**: `getOlder`, `makePurchase`, and `makeSpendingReport` define the behaviors or actions that a `Student` object can perform.

Python does not have explicit access modifiers like `public`, `private`, or `protected` found in some other languages. By convention, all attributes and methods are generally considered public. Developers often use a single leading underscore (`_attribute`) to suggest that an attribute is intended for internal use, and a double leading underscore (`__attribute`) triggers name mangling, making it harder to access directly from outside the class (though not impossible, it serves as a strong convention for 'private' members).

```python
class Student:
  def __init__(self, name, age):
    self.name = name
    self.age = age
    self.purchases = dict()

  def getOlder(self):
    self.age += 1

  def makePurchase(self, item: str, price: float):
    self.purchases[item] = price

  def makeSpendingReport(self):
    report = f"Spending Report for {self.name} (Age: {self.age}):\n"
    total_spent = 0.0
    if not self.purchases:
      report += "  No purchases made yet."
    else:
      for item, price in self.purchases.items():
        report += f"  - {item}: ${price:.2f}\n"
        total_spent += price
      report += f"Total spent: ${total_spent:.2f}"
    return report
```

### Using the `Student` Class

Here's an example of how to create a `Student` object, make some purchases, and generate a spending report.

```python
# Create a student instance
s = Student("Alice", 20)

# Make some purchases
s.makePurchase("Textbook", 75.50)
s.makePurchase("Laptop", 1200.00)
s.makePurchase("Coffee", 4.25)

# Get a spending report
print(s.makeSpendingReport())

# Make the student older
s.getOlder()
print(f"\n{s.name} is now {s.age} years old.")
print(s.makeSpendingReport()) # Report now shows updated age
```

```python
s = "banana"

# Get the first character
print(f"s[0]: {s[0]}")

# Get characters from index 1 to 4 (exclusive)
print(f"s[1:5]: {s[1:5]}")

# Get characters from the beginning to index 3 (exclusive)
print(f"s[:4]: {s[:4]}")

# Get characters from index 3 to the end
print(f"s[3:]: {s[3:]}")

# Get characters from the second to last to the end
print(f"s[-2:]: {s[-2:]}")

# Get characters in reverse order
print(f"s[::-1]: {s[::-1]}")

# Get every second character
print(f"s[::2]: {s[::2]}")
```
